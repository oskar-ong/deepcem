```python
from collections import defaultdict
import csv

new_pokemon_ids = defaultdict(int)
new_pokemon = []
pokemon_fp = "../data/raw/pokemon/50/pokemon.csv"
with open(pokemon_fp, "r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for i, row in enumerate(reader):
        original_id = row.get("pokemon", "").strip()
        row["pokemon"] = i
        new_pokemon_ids[original_id] = i
        # write 
        new_pokemon.append(row)

fieldnames = ["pokemon", "name", "species","height","weight","base_experience","is_default"]
with open("pokemon_new_ids.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(new_pokemon)
```

```python
# Duplicates
pokemon_dups = "../data/raw/pokemon/50/pokemon_dups.csv"
new_pokemon_dups_fp = "pokemon_dups_new_ids.csv"
new_pokemon_dups = []
with open(pokemon_dups, 'r') as infile:
    reader = csv.DictReader(infile)
    for row in reader:
        new_row = {"1": 0, "2": 0}
        original_id = row["1"]
        new_row["1"] = new_pokemon_ids[row["1"]]
        new_row["2"] = new_pokemon_ids[row["2"]]
        new_pokemon_dups.append(new_row)

fieldnames = ["1","2"]
with open(new_pokemon_dups_fp, 'w') as outfile:
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(new_pokemon_dups)
```

```python
# Related Tables
related_tables = ["poke_ability.csv", "poke_item.csv", "poke_move.csv"]

for t in related_tables:
    newname = f"new_{t}"
    fp = f"../data/raw/pokemon/50/{t}"
    new_rows = []
    with open(fp, "r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        fieldnames = reader.fieldnames
        for row in reader:
            row["pokemon"] = new_pokemon_ids[row["pokemon"]]
            new_rows.append(row)

    with open(newname, 'w') as outfile:
        writer = csv.DictWriter(outfile, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(new_rows)
    
```

```python
def write_new_ids(entity, in_fp, out_fp, prefix, related_tables):
    new_ids = {}
    new_rows = []
    with open(in_fp, "r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        fieldnames = reader.fieldnames
        for i, row in enumerate(reader):
            original_id = row.get(entity, "").strip()
            row[entity] = f"{prefix}{i}"
            new_ids[original_id] = f"{prefix}{i}"
            # write 
            new_rows.append(row)

    with open(out_fp, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(new_rows)


    dups = f"../data/raw/pokemon/50/{entity}_dups.csv"
    out = f"new_{entity}_dups.csv"
    new_rows = []
    with open(dups, 'r') as infile:
        reader = csv.DictReader(infile)
        fieldnames = reader.fieldnames
        for row in reader:
            new_row = {"1": 0, "2": 0}
            original_id = row["1"]
            new_row["1"] = new_ids[row["1"]]
            new_row["2"] = new_ids[row["2"]]
            new_rows.append(new_row)

    with open(out, 'w') as outfile:
        writer = csv.DictWriter(outfile, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(new_rows)

    for t in related_tables:
        # newname = f"new_{t}"
        # fp = f"../data/raw/pokemon/50/{t}"
        new_rows = []
        with open(t, "r", newline="", encoding="utf-8") as f:
            reader = csv.DictReader(f)
            fieldnames = reader.fieldnames
            for row in reader:
                row[entity] = new_pokemon_ids[row[entity]]
                new_rows.append(row)

        with open(t, 'w') as outfile:
            writer = csv.DictWriter(outfile, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(new_rows)
    
# ability

entity = "ability"
in_fp = f"../data/raw/pokemon/50/{entity}.csv"
out_fp = f"new_{entity}.csv"
prefix = "a"
related_tables = ["new_poke_ability.csv"]

write_new_ids(entity, in_fp, out_fp, prefix, related_tables)
```

```python
# item
entity = "item"
in_fp = f"../data/raw/pokemon/50/{entity}.csv"
out_fp = f"new_{entity}.csv"
prefix = "i"
related_tables = ["new_poke_item.csv"]

write_new_ids(entity, in_fp, out_fp, prefix, related_tables)
```

```python
# move 
entity = "move"
in_fp = f"../data/raw/pokemon/50/{entity}.csv"
out_fp = f"new_{entity}.csv"
prefix = "m"
related_tables = ["new_poke_move.csv"]

write_new_ids(entity, in_fp, out_fp, prefix, related_tables)
```
