```python
import os
import argparse
from collections import Counter

def count_labels(file_path):
    counts = Counter()
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            for _, line in enumerate(file, 1):
                clean_line = line.strip()
                if not clean_line:
                    continue
                
                last_val = clean_line[-1]
                if last_val in ('0', '1'):
                    counts[last_val] += 1
        return counts
    except FileNotFoundError:
        return None

def get_directory_details(dir_name):

    # The specific files we are looking for
    target_files = ["train.txt", "valid.txt", "test.txt", "train_125.txt"]
    
    print(f"{'File Name':<18} | {'0s':<6} | {'1s':<6} | {'Total':<6}")
    print("-" * 45)

    for filename in target_files:
        file_path = os.path.join(dir_name, filename)
        
        if os.path.exists(file_path):
            results = count_labels(file_path)
            zeros = results['0']
            ones = results['1']
            total = zeros + ones
            print(f"{filename:<18} | {total} & {ones} & {zeros} ")
        else:
            print(f"{filename:<18} | [File Not Found]")

get_directory_details("../data/pokemon/source/baseA/species")

```
