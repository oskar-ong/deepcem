# Analyze Confidence Scores

```python
import json
import re

def compare_ditto_results(jsonl_path, txt_path):
    """
    Compares Ditto JSONL predictions with the actual text labels line-by-line.
    
    Returns:
        List of tuples: (line_number, predicted_class, actual_class, match_confidence)
    """
    comparison_results = []
    
    # Matches the final integer at the very end of the line
    actual_label_pattern = re.compile(r'(\d+)\s*$')

    with open(jsonl_path, 'r', encoding='utf-8') as f_pred, \
         open(txt_path, 'r', encoding='utf-8') as f_actual:
        
        for line_num, (line_pred, line_actual) in enumerate(zip(f_pred, f_actual), 1):
            # 1. Parse JSONL predictions
            data_pred = json.loads(line_pred.strip())
            predicted_class = data_pred.get('match')
            match_confidence = data_pred.get('match_confidence')
            
            # 2. Extract actual class from the text file line
            match_actual = actual_label_pattern.search(line_actual)
            if match_actual:
                actual_class = int(match_actual.group(1))
            else:
                actual_class = None  # Fallback if a line lacks a trailing integer
            
            # 3. Store the structured tuple
            if actual_class != predicted_class:
                comparison_results.append((
                    line_num, 
                    predicted_class, 
                    actual_class, 
                    match_confidence
                ))
            
    return comparison_results

# --- Run the comparison ---
if __name__ == "__main__":
    # file1 = "../imdb_results/movie_3_conv_results.jsonl"
    file1 = "../imdb_results/movie_baseline.jsonl"
    file2 = "../data/imdb/high/baseA/movie/test.txt"
    
    results = compare_ditto_results(file1, file2)
    
    # Quick look at the output structure
    print(f"Processed {len(results)} lines successfully.\n")
    print("Sample output tuple format:")
    print("(line_number, predicted, actual, confidence)")
    print("-" * 45)
    for entry in results:  # Adjust to see more rows if needed
        print(entry)
```

```python
import json
import re
import statistics

def analyze_model_improvements(file1_path, file2_path, file3_path):
    actual_label_pattern = re.compile(r'(\d+)\s*$')
    
    dataset = []
    
    with open(file1_path, 'r', encoding='utf-8') as f_base, \
         open(file2_path, 'r', encoding='utf-8') as f_actual, \
         open(file3_path, 'r', encoding='utf-8') as f_new:
        
        for line_base, line_actual, line_new in zip(f_base, f_actual, f_new):
            data_base = json.loads(line_base.strip())
            data_new = json.loads(line_new.strip())
            
            base_pred = data_base.get('match')
            base_conf = data_base.get('match_confidence')
            new_pred = data_new.get('match')
            
            match_actual = actual_label_pattern.search(line_actual)
            actual_class = int(match_actual.group(1)) if match_actual else None
            
            if actual_class is not None and base_conf is not None:
                dataset.append({
                    'base_conf': base_conf,
                    'base_correct': (base_pred == actual_class),
                    'new_correct': (new_pred == actual_class)
                })

    total_records = len(dataset)
    if total_records == 0:
        print("No valid data found to analyze.")
        return

    thresholds = [0.4, 0.5, 0.70, 0.80, 0.85, 0.90, 0.95, 0.97, 0.99, 0.995]
    
    print(f"{total_records} Total Pairs")
    print(f"{'Baseline Threshold':<20} | {'Remaining Errors'}")
    print("-" * 75)
    
    for t in thresholds:
        # Traffic that stays with baseline because confidence >= t
        baseline_traffic = [row for row in dataset if row['base_conf'] >= t]
        
        if not baseline_traffic:
            continue
            
        cost_saved_pct = (len(baseline_traffic) / total_records) * 100
        correct_count = sum(1 for row in baseline_traffic if row['base_correct'])
        baseline_accuracy = (correct_count / len(baseline_traffic)) * 100
        errors_passed = len(baseline_traffic) - correct_count
        
        # print(f"Confidence >= {t:<7.3f} | {cost_saved_pct:<13.1f}% | {baseline_accuracy:<18.2f}% | {errors_passed:<14}")
        print(f"Confidence >= {t:<7.3f} | {errors_passed:<14}")


file1 = "../imdb_results/movie_baseline.jsonl"
file2 = "../data/imdb/high/baseA/movie/test.txt"
file3 = "../imdb_results/movie_3_conv_results.jsonl"

analyze_model_improvements(file1, file2, file3)
    
```
